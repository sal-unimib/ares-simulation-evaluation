```python
# RQ1 - Constraint Satisfaction Rate / Compliance
import pandas as pd
import plotly.graph_objects as go

csr_df = pd.read_csv('data/evaluation/rq1/csr.csv')
csr_df = csr_df[csr_df['scenario'] != 'stable']

fig = go.Figure()

fig.add_trace(go.Bar(
    x=csr_df['scenario'],
    y=csr_df['baseline'],
    name='Baseline',
    text=[f"{v:.2f}" for v in csr_df['baseline']],
    textposition='outside'
))

fig.add_trace(go.Bar(
    x=csr_df['scenario'],
    y=csr_df['ares'],
    name='ARES',
    text=[f"{v:.2f}" for v in csr_df['ares']],
    textposition='outside'
))

fig.update_layout(
    yaxis_title='CSR',
    barmode='group',
    width=600,
    height=600
)

fig.update_layout()

fig.show()
```

```python
# RQ 2 - Effectiveness
import pandas as pd
import plotly.graph_objects as go

df_deg = pd.read_csv('data/evaluation/rq2/effectiveness_degradation.csv')
df_dyn = pd.read_csv('data/evaluation/rq2/effectiveness_dynamic.csv')

def get_values(df):
    baseline = df[df['system'] == 'Baseline']['effectiveness'].values[0]
    ares = df[df['system'] == 'ARES']['effectiveness'].values[0]
    return baseline, ares

baseline_deg, ares_deg = get_values(df_deg)
baseline_dyn, ares_dyn = get_values(df_dyn)

scenarios = ['degradation', 'dynamic']

fig = go.Figure()

# Baseline
fig.add_trace(go.Bar(
    x=scenarios,
    y=[baseline_deg, baseline_dyn],
    name='Baseline',
    text=[f"{baseline_deg:.2f}", f"{baseline_dyn:.2f}"],
    textposition='outside'
))

# ARES
fig.add_trace(go.Bar(
    x=scenarios,
    y=[ares_deg, ares_dyn],
    name='ARES',
    text=[f"{ares_deg:.2f}", f"{ares_dyn:.2f}"],
    textposition='outside'
))

fig.update_layout(
    yaxis_title='Effectiveness',
    barmode='group',
    yaxis=dict(range=[0, 0.3]),
    width=600,
    height=600
)

fig.show()
```

```python
# RQ3 - CSR - Cost

import pandas as pd
import plotly.graph_objects as go

csr = pd.read_csv('data/evaluation/rq1/csr.csv')
cost = pd.read_csv('data/evaluation/rq3/cost.csv')

csr = csr[csr['scenario'] != 'stable']
cost = cost[cost['scenario'] != 'stable']

fig = go.Figure()

for scenario in csr['scenario']:
    csr_baseline = csr[csr['scenario'] == scenario]['baseline'].values[0]
    csr_ares = csr[csr['scenario'] == scenario]['ares'].values[0]
    
    cost_baseline = cost[cost['scenario'] == scenario]['baseline'].values[0]
    cost_ares = cost[cost['scenario'] == scenario]['ares'].values[0]
    
    # Baseline
    fig.add_trace(go.Scatter(
        x=[cost_baseline],
        y=[csr_baseline],
        mode='markers+text',
        name='Baseline',
        text=[scenario],
        textposition='top left',
        marker=dict(size=12, color='blue')
    ))
    
    # ARES
    fig.add_trace(go.Scatter(
        x=[cost_ares],
        y=[csr_ares],
        mode='markers+text',
        name='ARES',
        text=[scenario],
        textposition='top right',
        marker=dict(size=12, color='red')
    ))

fig.update_layout(
    xaxis_title='Cost',
    yaxis_title='CSR',
    width=600,
    height=600
)

fig.show()
```
